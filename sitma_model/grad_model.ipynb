```python
import tensorflow as tf
```

```python
eski_model_yolu = '/Users/sametkarayel/Desktop/sitma_tespiti/sitma_model/sitma_tespit_modeli.keras'
base_model = tf.keras.models.load_model(eski_model_yolu)
```

```python
inputs = tf.keras.Input(shape=(128, 128, 3))
x = inputs
for layer in base_model.layers:
    # InputLayer'ı atla, biz zaten kendi Input'umuzu verdik
    if not isinstance(layer, tf.keras.layers.InputLayer):
        x = layer(x)
```

```python
grad_model =tf.keras.Model(inputs=inputs,outputs=x)
```

```python
grad_model.save('grad_model.keras')
```
